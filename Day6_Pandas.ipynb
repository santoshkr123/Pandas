```python
import pandas as pd

data = {
    "Name": ["Rahul", "Priya", "Amit", "Neha", "Rohit", "Sneha", "Vikas", "Pooja"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT", "Finance", "HR"],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female", "Male", "Female"],
    "Salary": [50000, 45000, 65000, 55000, 40000, 70000, 60000, 50000],
    "Experience": [2, 5, 7, 4, 1, 6, 5, 3]
}

df = pd.DataFrame(data)
df
```

```python
#What is the average salary for each Department and Gender?
df.groupby(["Department","Gender"])["Salary"].mean()
```

```python
#Show departments from highest average salary to lowest. Sorting Grouped Results
result=(
    df.groupby("Department")["Salary"].mean().reset_index().sort_values("Salary",ascending=False)
)
result
```

```python
#Filtering Groups .Find departments where average salary is greater than 50,000.
avg_sal=df.groupby("Department")["Salary"].mean()
avg_sal[avg_sal>50000]
```

```python
#groupby().filter() -Keep only departments having more than 2 employees.
df.groupby("Department").filter(
    lambda x:len(x) > 2
)
```

```python
#transform()-Add the average salary of each employee's department.
df["avg_sal_transform"]=df.groupby("Department")["Salary"].transform("mean")


df
```

```python
#Compare Employee Salary With Department Average
#Which employees earn more than their department's average salary?

df["avg_sal_dep"]=df.groupby("Department")["Salary"].transform("mean")
df[df["Salary"] > df["avg_sal_dep"]]
```

```python
#Calculate Salary Difference
df["salary_diff"]=df["Salary"]- df["avg_sal_dep"]
df
```

```python
#Percentage Difference
df["per_diff"]=(df["Salary"]-df["avg_sal_dep"]) / df["avg_sal_dep"]*100
df
```

```python
#Find Highest-Paid Employee in Each Department
idx=df.loc[df.groupby("Department")["Salary"].idxmax()]
idx
```

```python
#Group-Wise Ranking - rank employees within their department.
df["department_rank_emp"]=df.groupby("Department")["Salary"].rank(ascending=False)
df
```

```python
import pandas as pd

data = {
    "Name": ["Rahul", "Priya", "Amit", "Neha", "Rohit", "Sneha", "Vikas", "Pooja"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT", "Finance", "HR"],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female", "Male", "Female"],
    "Salary": [50000, 45000, 65000, 55000, 40000, 70000, 60000, 50000],
    "Experience": [2, 5, 7, 4, 1, 6, 5, 3]
}

df = pd.DataFrame(data)
df
```

```python
Q1. Find average salary for each Department + Gender.

Q2. Find maximum salary for each Department + Gender.

Q3. Find employee count for each Department + Gender.

Q4. Find total salary for each department.

Q5. Sort departments by average salary from highest to lowest.
```

```python
#Q6. Find departments having more than 2 employees.
df.groupby("Department").filter(
    lambda x :len(x)>2
)

```

```python
#Q9. Find employees belonging to departments with more than 2 employees.
df.groupby("Department").filter(
    lambda x :len(x) > 2
)
```

```python
#Q10. Find employees whose salary is greater than their department's average salary.
df["avg_sal"]=df.groupby("Department")["Salary"].transform("mean")
df[df["Salary"]>df["avg_sal"]]
```

```python
Q11. Create a column:

Department_Avg_Salary

Q12. Create:

Salary_Difference

Q13. Create:

Salary_Difference_Percent

Q14. Find employees earning below their department average.

Q15. Find employees earning more than 10% above their department average.
```

```python
#Q11. Create a column: Department_Avg_Salary
df["Department_Avg_Salary"]=df.groupby("Department")["Salary"].transform("mean")
df
```

```python
#Q12. Create:Salary_Difference
df["Salary_Difference"]=df["Salary"]-df["Department_Avg_Salary"]
df
```

```python
#Q13. Create:Salary_Difference_Percentage
df["Salary_Difference_Percentage"]=df["Salary"]-df["Department_Avg_Salary"]/df["Department_Avg_Salary"]
df
```

```python
#Q14. Find employees earning below their department average.
df["dept_avg"]=df.groupby("Department")["Salary"].transform("mean")
df[df["Salary"] < df["dept_avg"]]
```

```python
#Q15. Find employees earning more than 10% above their department average.
df["dept_avg"]=df.groupby("Department")["Salary"].transform("mean")
df[df["Salary"]>df["dept_avg"]*0.10]
```
