```python
import pandas as pd

data = {
    "Name": ["Rahul", "Priya", "Amit", "Neha", "Rohit", "Sneha"],
    "Age": [25, 28, 30, 26, 24, 29],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Salary": [50000, 45000, 65000, 55000, 40000, 70000],
    "Experience": [2, 5, 7, 4, 1, 6]
}

df = pd.DataFrame(data)
df
```

```python
#Suppose we want to sort employees by salary.
df.sort_values("Salary")
```

```python
#Suppose we want to sort employees by salary.-highest salary
df.sort_values("Salary",ascending=False)
```

```python
#Sort and select Top N -Find the highest-paid employee.

df.sort_values("Salary",ascending=False).head(1)
```

```python
#Top 3 employees
df.sort_values("Salary",ascending=False).head(3)
```

```python
#nlargest() - 
df.nlargest(3,"Salary")
```

```python
#nsmallest()
df.nsmallest(3,"Salary")
```

```python
#Sorting by Multiple Columns
'''Suppose we want to sort:

Department alphabetically
Salary from highest to lowest within each department'''

df.sort_values(by=["Department","Salary"],ascending=[True,False])
```

```python
#rank()
df["Salary_Rank"]=df["Salary"].rank(ascending=False)
df
```

```python
df["Salary"].rank(method="dense",ascending=False)
```

```python
#filtering and sorting
#Find IT employees and sort them by salary from highest to lowest.
df[df["Department"]=="IT"].sort_values(
    "Salary",ascending=False)
```

```python
#Filter + Sort + Top N -Find the top 2 highest-paid IT employees.
df[df["Department"]=="IT"].sort_values("Salary",ascending=False).head(2)
```

```python
#Find the top 3 HR employees by salary.
df[df["Department"]=="HR"].nlargest(3,"Salary")
```

```python
import pandas as pd

data = {
    "Name": ["Rahul", "Priya", "Amit", "Neha", "Rohit", "Sneha"],
    "Age": [25, 28, 30, 26, 24, 29],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Salary": [50000, 45000, 65000, 55000, 40000, 70000],
    "Experience": [2, 5, 7, 4, 1, 6]
}

d = pd.DataFrame(data)
d
```

```python
#Q1. Sort the DataFrame by Salary from lowest to highest.
d.sort_values("Salary",ascending=True)
```

```python
#Q2. Sort the DataFrame by Salary from highest to lowest.
d.sort_values("Salary",ascending=False)
```

```python
#Q3. Find the employee with the highest salary using sort_values().
d.sort_values("Salary",ascending=False)
```

```python
#Q4. Find the employee with the lowest salary.
d.sort_values("Salary",ascending=True)
```

```python
#Q5. Find the top 3 highest-paid employees using nlargest().
d.nlargest(3,"Salary")
```

```python
#Q6. Find the bottom 2 employees by salary using nsmallest().
d.nsmallest(2,"Salary")
```

```python
#Q7. Sort employees by Age from youngest to oldest.
d.sort_values("Age",ascending=True)
```

```python
#Q8.Sort employees by Experience from highest to lowest.
d.sort_values("Experience",ascending=False)
```

```python
#Q9. Sort by Department ascending and Salary descending.
d.sort_values(by=["Department","Salary"],ascending=[True,False])
```

```python
#Q10. Find the top 2 highest-paid employees in the IT department.
d[d["Department"]=="IT"].nlargest(2,"Salary")
```

```python
#Q11. Find the bottom 2 lowest-paid employees in the HR department.
d[d["Department"]=="HR"].nsmallest(2,"Salary")
```

```python
#Q12. Find employees whose salary is greater than 45000, then sort them by salary descending.
d[d["Salary"]>45000].sort_values("Salary",ascending=False)
```

```python
#Q13. Create a new column called Salary_Rank where the highest salary gets rank 1.
d["Salary_Rank"]=d["Salary"].rank(ascending=False)
d
```

```python
#Q14. Find the top 3 employees based on Experience.
d.nlargest(3,"Experience")
```

```python
#Q15. Find the highest-paid employee in the IT department.
d[d["Department"]=="IT"].sort_values("Salary",ascending=False)
```

```python
#Q16. Find the second-highest salary.
d.nlargest(2,"Salary")
```

```python
#Q17. Find the second-highest-paid employee.
d.nlargest(2,"Salary")
```

```python
#Q18. Find the top 2 employees by salary among employees with more than 3 years of experience.
d[d["Experience"] >3].nlargest(2,"Salary")
```

```python
#Q19. Sort employees by Department ascending, and within each department sort Experience from highest to lowest.
d.sort_values(by=["Department","Experience"],ascending=[True,False])
```

```python
#Q20. 🔥 Interview Problem:Find the top 2 highest-paid employees from each department.
d.groupby("Department").apply(lambda x:x.nlargest(2,"Salary"))
```

```python
#Q21 🔥Find the top 2 most experienced employees from each department.
d.groupby("Department").apply(lambda x:x.nlargest(2,"Experience"))
```
