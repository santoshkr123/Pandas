```python
import pandas as pd
employees = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Rahul", "Priya", "Amit", "Neha", "Rohit"],
    "Department_ID": [1, 2, 1, 3, 4],
    "Salary": [50000, 45000, 65000, 55000, 40000]
})

departments = pd.DataFrame({
    "Department_ID": [1, 2, 3],
    "Department": ["IT", "HR", "Finance"]
})

sales = pd.DataFrame({
    "Order_ID": [1001, 1002, 1003, 1004, 1005],
    "Customer_ID": [1, 2, 1, 3, 5],
    "Amount": [500, 700, 300, 900, 400]
})#

customers = pd.DataFrame({
    "Customer_ID": [1, 2, 3, 4],
    "Customer_Name": ["Rahul", "Priya", "Amit", "Neha"]
})
orders = pd.DataFrame({
    "Order_ID": [101, 102, 103, 104, 105],
    "Customer_ID": [1, 2, 1, 3, 5],
    "Amount": [500, 700, 300, 900, 400]
})
salary = pd.DataFrame({
    "Salary": [50000, 45000, 65000]
}, index=[101, 102, 103])
```

```python
Q1. Concatenate two DataFrames vertically.

Q2. Concatenate two DataFrames horizontally.

Q3. Merge employees and departments using Department_ID.

Q4. Perform an inner join.

Q5. Perform a left join.
```

```python
#Q1. Concatenate two DataFrames vertically.
pd.concat([employees,departments],axis=0)
```

```python
#Q2. Concatenate two DataFrames horizontally.
pd.concat([employees,departments],axis=1)
```

```python
#Q3. Merge employees and departments using Department_ID.
pd.merge(employees ,departments,on="Department_ID")
```

```python
#Q4. Perform an inner join.
pd.merge(employees ,departments,on="Department_ID",how="inner")
```

```python
#Q5. Perform a left join.
pd.merge(employees ,departments,on="Department_ID",how="left")
```

```python
Q6. Perform a right join.

Q7. Perform an outer join.

Q8. Merge sales and customers.

Q9. Keep all sales records while adding customer names.

Q10. Find sales records where the customer does not exist in the customer table.
```

```python
#Q6. Perform a right join.
pd.merge(employees ,departments,on="Department_ID",how="right")
```

```python
#Q7. Perform an outer join.
pd.merge(employees ,departments,on="Department_ID",how="outer")
```

```python
#Q8. Merge sales and customers.
pd.merge(sales,customers)
```

```python
#Q9. Keep all sales records while adding customer names.
pd.merge(sales,customers,on="Customer_ID",how="left")
```

```python
#Q10. Find sales records where the customer does not exist in the customer table.
miss_cust=pd.merge(sales,customers,on="Customer_ID",how="left",indicator=True)
res=miss_cust[miss_cust["_merge"]=="left_only"]
res
```

```python
departments2 = pd.DataFrame({
    "Dept_ID": [1, 2, 3],
    "Department": ["IT", "HR", "Finance"]
})
departments2
```

```python
Q11. Merge employees with departments2 using Department_ID and Dept_ID.

Q12. Keep all employees.

Q13. Keep all departments.
```

```python
#Q11. Merge employees with departments2 using Department_ID and Dept_ID.
pd.merge( employees, departments2 ,
        left_on="Department_ID" ,
        right_on="Dept_ID")
```

```python
#Q12. Keep all employees
pd.merge(employees, departments2 ,
        left_on="Department_ID" ,
        right_on="Dept_ID",
        how="left")
```

```python
#Q13. Keep all departments.
pd.merge(employees, departments2 ,
        left_on="Department_ID" ,
        right_on="Dept_ID",
        how="right")
```

```python
Q14. Find employees whose department information is missing.

Q15. Find departments that have no employees.

Q16. Create a final employee dataset containing:

Employee_ID
Name
Department
Salary

Q17. Calculate average salary by department using the merged dataset.

Q18. Find the department with the highest average salary.

Q19. Find employees whose department is missing.

Q20. Find customers who have never placed an order.
```

```python
#Q14. Find employees whose department information is missing.
missing_department=pd.merge(employees,
        departments,
        on="Department_ID",
        how="left",indicator=True)
res=missing_department[missing_department["_merge"]=="left_only"]

res
```

```python
#Q15. Find departments that have no employees
missing_emp=pd.merge(
        departments,employees,
        on="Department_ID",
        how="left",
indicator=True)

res=missing_emp[missing_emp["_merge"]=="left_only"]
res
```

```python
#Q16. Create a final employee dataset containing:

'''Q16. Create a final employee dataset containing:

Employee_ID
Name
Department
Salary'''

final_res=pd.merge(employees,departments,on="Department_ID",how="left")
res=final_res[["Employee_ID","Name","Department","Salary"]]
res
```

```python
#Q17. Calculate average salary by department using the merged dataset.
avg_sal=pd.merge(employees,departments,on="Department_ID",how="left")
res=avg_sal.groupby("Department")["Salary"].mean().reset_index()
res
```

```python
#Q18. Find the department with the highest average salary.
high_avg_sal=pd.merge(employees,departments,on="Department_ID",how="left")
high_avg_sal.groupby("Department")["Salary"].mean().idxmax()
```

```python
#Q19. Find employees whose department is missing.
missing_Dept=pd.merge(employees,departments,on="Department_ID",how="left",indicator=True)
res=missing_Dept[missing_Dept["_merge"]=="left_only"]
res
```

```python
#Q20. Find customers who have never placed an order.
result=pd.merge(customers,sales,on="Customer_ID",how="left",indicator=True)
result[result["_merge"]=="left_only"]
```

```python

'''Q21.

Find unmatched employees after joining employees with departments.'''

result=pd.merge(employees,departments,on="Department_ID",how="left",indicator=True)
result[result["_merge"]=="left_only"]
```

```python
#Find unmatched departments after joining departments with employees.
res=pd.merge(employees,departments,on="Department_ID",how="right",indicator=True)
res[res["_merge"]=="right_only"]
```

```python
'''You have an orders table and a customers table.
You must preserve every order, even if customer information is missing. 
Which join would you use and why?'''
res=pd.merge(orders,customers,on="Customer_ID",how="left",indicator=True)
res[res["_merge"]=="left_only"]
```
