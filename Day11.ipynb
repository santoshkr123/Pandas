```python
import pandas as pd
import numpy as np

data = {
    "Name": [
        "Rahul", "Priya", "Amit", "Neha",
        "Rohit", "Sneha", "Vikas", "Pooja"
    ],
    "Age": [25, 32, 28, 41, 24, 35, 29, 45],
    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "IT", "Finance", "HR"
    ],
    "Salary": [
        50000, 65000, 45000, 80000,
        40000, 70000, 60000, 55000
    ],
    "Experience": [
        2, 6, 3, 10, 1, 7, 5, 8
    ]
}

df = pd.DataFrame(data)
df
```

```python
Q1

Create Annual_Salary.

Salary × 12

Use assign().

Q2

Create:

Annual_Salary
Annual_Bonus

where:

Annual_Bonus = Annual_Salary × 10%

Use one assign().

Q3

Create:

Experience_Months

using:

Experience × 12
Q4

Create:

Salary_Lakh

Formula:

Salary / 100000
Q5

Using assign(), create both:

Annual_Salary
Salary_Lakh
```

```python
'''Q1

Create Annual_Salary.

Salary × 12

Use assign().'''
df.assign(
    Annula_Salary=df["Salary"]*12
)
```

```python
'''Q2

Create:

Annual_Salary
Annual_Bonus

where:

Annual_Bonus = Annual_Salary × 10%'''

df.assign(
    Annual_Salary=df["Salary"]*12,
    Annual_Bonus=lambda x :x["Annual_Salary"]*0.10
)
```

```python
'''Q3

Create:

Experience_Months

using:

Experience × 12'''
df.assign(
    Experience_Months=df["Experience"]*12
)
```

```python
'''Q4

Create:

Salary_Lakh

Formula:

Salary / 100000'''
df.assign(
    Salary_Lakh=df["Salary"]/100000
)
```

```python
Q11

Using where(), keep salaries >= 50000 and replace the others with 0.

Q12

Using mask(), replace salaries < 50000 with 0.

Q13

Using where(), keep employees whose experience is >= 5; replace others with NaN.

Q14

Using mask(), replace ages greater than 40 with NaN.

Q15 💡

Create:

Salary_Status

using where() so that:

Salary >= 60000 → original Salary
Salary < 60000  → 0
```

```python
'''Q11

Using where(), keep salaries >= 50000 and replace the others with 0.
'''
df.where(
    df["Salary"]>=50000,
    0
)
```

```python
'''Q12

Using mask(), replace salaries < 50000 with 0.'''
df.where(
    df["Salary"]<50000,
    0
)
```

```python
'''Q13

Using where(), keep employees whose experience is >= 5; replace others with NaN.'''
df.where(
    df["Experience"]>=5
)
```

```python
'''14

Using mask(), replace ages greater than 40 with NaN.
'''
df.mask(
    df["Age"]>40
)
```

```python
'''
Q15 💡

Create:

Salary_Status

using where() so that:

Salary >= 60000 → original Salary
Salary < 60000  → 0'''

df["Salary_Status"]=np.where(
    df["Salary"]>=60000,"original Salary",0
)
df
```

```python
Q16

Cap Salary at:

70000

Create:

Salary_Capped
Q17

Set a salary range:

minimum = 45000
maximum = 75000

using clip().

Q18

Create:

Age_Capped

where age cannot exceed 40.

Q19

Create:

Experience_Capped

where experience cannot exceed 7.

Q20 💼

Explain why blindly using clip() on a suspicious value can be dangerous in real-world data analysis.

Think about:

Data validation
vs
Data manipulation
```

```python
'''16

Cap Salary at:

70000

Create:

Salary_Capped'''
df["Salary_Capped"]=df["Salary"].clip(
    upper=70000
)
df
```

```python
'''Q17

Set a salary range:

minimum = 45000
maximum = 75000

using clip().'''

df["min_and_max_caped"]=df["Salary"].clip(
    lower=45000,
    upper=75000
)
df
```

```python
'''Q18

Create:

Age_Capped

where age cannot exceed 40.
'''
df["Age_caped"]=df["Age"].clip(
    upper=40
)
df
```

```python
'''Q19

Create:

Experience_Capped

where experience cannot exceed 7.
'''
df["Experience_Capped"]=df["Experience"].clip(
    upper=7
)
df
```
