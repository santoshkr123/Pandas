```python
deliveries = pd.DataFrame({
    "Order_ID": [101, 102, 103, 104, 105],
    "Order_Date": [
        "2025-01-01",
        "2025-01-10",
        "2025-02-05",
        "2025-02-20",
        "2025-03-01"
    ],
    "Delivery_Date": [
        "2025-01-05",
        "2025-01-15",
        "2025-02-08",
        "2025-02-25",
        "2025-03-10"
    ]
})
deliveries 
```

```python
Q15. Convert both columns to datetime.

Q16. Calculate Delivery_Days.

Q17. Find the average delivery time.

Q18. Find the order with the longest delivery time.

Q19. Find orders delivered within 5 days.
```

```python
#Q15. Convert both columns to datetime.
deliveries["Order_Date"]=pd.to_datetime(deliveries["Order_Date"])
deliveries["Delivery_Date"]=pd.to_datetime(deliveries["Delivery_Date"])
```

```python
deliveries
```

```python
deliveries.dtypes
```

```python
#Q16. Calculate Delivery_Days.
deliveries["Delivery_time"]=deliveries["Delivery_Date"]-deliveries["Order_Date"]
deliveries
```

```python
#Q17. Find the average delivery time.

deliveries["AVG_Delivery"]=deliveries["Delivery_time"].mean()
deliveries
```

```python
#Q18. Find the order with the longest delivery time.
deliveries["Longest_delivery_time"]=deliveries["Delivery_time"].max()
deliveries
```

```python
#Q19. Find orders delivered within 5 days.
len(deliveries[deliveries["Delivery_time"] <=pd.Timedelta(days=5)])
```
