```python
import pandas as pd
import numpy as np

data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    
    "Name": [
        "Rahul", "Priya", "Amit", "Neha", "Rohit",
        "Sneha", "Vikas", "Pooja", "Karan", "Anjali"
    ],
    
    "Age": [
        25, 32, np.nan, 41, 24,
        35, np.nan, 45, 31, 27
    ],
    
    "Department": [
        "IT", "HR", "IT", "Finance", "HR",
        "IT", "Finance", "HR", "IT", "Finance"
    ],
    
    "Salary": [
        50000, 65000, np.nan, 80000, 40000,
        70000, 60000, np.nan, 75000, 48000
    ],
    
    "Experience": [
        2, 6, 3, np.nan, 1,
        7, 5, 8, np.nan, 2
    ],
    
    "City": [
        "Delhi", "Mumbai", "Delhi", None, "Mumbai",
        "Delhi", "Pune", "Mumbai", "Delhi", None
    ]
}

df = pd.DataFrame(data)
df
```

```python
isna()-This returns True where a value is missing. detect missing values
```

```python
df.isna()
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df.isnull().sum().sum() #total missing values
```

```python
df[df.isnull().any(axis=1)] #find rows containing missing values
```

```python
df[df.notna().all(axis=1)]#find rows containing no missing values
```

```python
#filling missing values with mean
df["Age"]=df["Age"].fillna(
    df["Age"].mean()
)
df
```

```python
df["Age"]=df["Age"].fillna(
    df["Age"].median()
)
df
```

```python
df["City"]=df["City"].fillna(
    df["City"].mode()[0]
)
df
```

```python
df["City"]=df["City"].fillna("Unknown")
df
```

```python
df.dropna(subset=["Salary"]) #drop rows when speciifc column is missing
```

```python
df.dropna()#remove rows containing missing values
```

```python
df["Salary"].ffill() #It uses the previous available value.
```

```python
df["Salary"].bfill() #It uses the next available value.
```

```python
#replace
df["City"]=df["City"].replace(
    ["NA","N/A","-",""],np.nan
)
df
```

```python
#pd.to_numeric()
df["Salary"]=pd.to_numeric(
    df["Salary"],
    errors="coerce"
)
df
```

```python
pd.to_datetime(
    df["joining_Date"],
    errors="coerce"
)
```

```python
df.duplicated() #detect dupicate rows
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates() #remove duplicate rows
```

```python
df.duplicated(
    subset=["Employee_ID"]
)
```

```python
df.drop_duplicates(
    subset=["Employee_ID"],
    keep="first"
)
```

```python
#group by wise missing values
```

```python
df["Salary"]=df.groupby("Department")["Salary"].fillna(
    df["Salary"].transform(
        lambda x:x.fillna(x.mean())
    )
)
df
```
