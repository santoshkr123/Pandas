```python
'''Q6

Create Age_Group:

Age < 30       → Young
30–39          → Adult
40+            → Senior'''

def age_cat(age):
    if age<30:
        return "Young"
    elif age <40:
        return "Adult"
    else:
        return "Senior"
```

```python
df["Age_Group"]=df["Age"].apply(age_cat)
df
```

```python
'''Q7

Create Salary_Category:

Salary < 50000      → Low
50000–69999         → Medium
70000+              → High'''

def sal_cat(salary):
    if salary<50000:
        return "Low"
    elif salary<70000:
        return "Medium"
    else:
        return "High"
```

```python
df["Salary_Category"]=df["Salary"].apply(sal_cat)
df
```

```python
'''Q8

Create Experience_Level:

Experience < 3      → Beginner
3–6                 → Intermediate
7+                  → Experienced'''

def exp_level(exp):
    if exp<3:
        return "Beiginner"
    elif exp<7:
        return "Intermediate"
    else:
        return "Experienced"
```

```python
'''Q9

Create a Tax column:

Salary >= 70000 → 20% tax
Salary < 70000  → 10% tax'''
import numpy as np

df["Tax"]=df["Salary"]* np.where(df["Salary"]>=70000 ,0.20,0.10)
df
```

```python
'''Q16.Create Bonus using this rule:

Experience >= 5 AND Salary >= 60000
        → 15% bonus

Otherwise
        → 5% bonus'''

def bonus(row):
    if row["Experience"]>=5 and row["Salary"]>=60000:
        return row["Salary"]*0.15
    else:    
        return row["Salary"]*0.05
```

```python
df["Bonus"]=df.apply(bonus,axis=1)
df
```

```python
'''Q17

Create Employee_Level:

Experience >= 7 → Senior
Experience >= 3 → Mid
Otherwise        → Junior'''

def emp_level(emp):
    if emp>=7:
        return "Senior"
    elif emp>=3:
        return "Mid"
    else:
        return "Junior"
```

```python
df["Employee_Level"]=df["Experience"].apply(emp_level)
df
```

```python
'''Q18

Create Risk_Category.

Rules:

Age > 40 AND Salary < 60000
    → High Risk

Age > 30 AND Salary < 60000
    → Medium Risk

Otherwise
    → Low Risk'''

def risk_category(age):
    if age["Age"]>40 and age["Salary"]<60000:
        return "High Risk"
    elif age["Age"]>30 and age["Salary"]<60000:
        return "Medium Risk"
    else:
        "Low Risk"
```

```python
df["Risk_Category"]=df.apply(risk_category,axis=1)
df
```

```python
'''Q19

Create:

Salary_Per_Year_Experience

Formula:

Salary / Experience

Handle the possibility of Experience = 0.

'''

df["Salary_Per_Year_Experience"]=df["Salary"]/df["Experience"].replace(0,np.nan)
df
```

```python
'''Q20

Create:

Employee_Score

Formula:

Salary / 10000 + Experience'''

df["Employee_Score"]=df["Salary"]/1000+df["Experience"]
df
```

```python
'''Q24 — Interview Problem 💼

A company wants to identify employees eligible for a special bonus.

Conditions:

Experience >= 5
AND
Salary >= 60000
AND
Department == "IT"

Create:

Special_Bonus

Rules:

Eligible     → 20000
Not Eligible → 0'''

def special_bonus(row):
    if row["Experience"]>=5 and row["Salary"] >=60000 and row["Department"]=="IT":
        return 20000
    else:
        return 0


df["Special_Bonus"]=df.apply(special_bonus,axis=1)
df
```

```python
'''Q22

Create:

Promotion_Status

Rules:

Experience >= 7 AND Salary >= 60000
        → Promote

Experience >= 5
        → Review

Otherwise
        → Not Ready'''


def promo(row):
    if row["Experience"]>=7 and row["Salary"]>=60000:
        return "Promote"
    elif row["Experience"] >=5:
        return "Review"
    else:
        "Not Ready"
df["Promotion_Status"]=df.apply(promo,axis=1)        
df
```

```python
'''Q21

Create:

Employee_Category

Rules:

Salary >= 70000 AND Experience >= 7
        → Top Performer

Salary >= 50000 AND Experience >= 3
        → Good Performer

Otherwise
        → Developing'''

def emp_cat(row):
    if row["Salary"]>=70000 and row["Experience"]>=7:
        return "Top Performer"
    elif row["Salary"]>=50000 and row["Experience"]>=3:
        return "Good Performer"
    else:
        "Developing"
df["Employee_Category"]=df.apply(emp_cat,axis=1)    
df
```

```python
'''Q23

Create:

Employee_Profile

Format:

Rahul - IT - 50000
Priya - HR - 65000
...

Hint:

You need multiple columns.'''

df["Employee_Profile"]=df.apply(
    lambda row:f"{row["Name"]} - {row["Department"]} - {row["Salary"]}",axis=1
)
df
```

```python
'''Create a new column:

Employee_Segment

Rules:

Top
Salary >= 70000
AND
Experience >= 7
Mid
Salary >= 50000
AND
Experience >= 3
Entry

Everything else.

Then answer:

How many employees belong to each segment?'''
```

```python
def emp_seg(row):
    if row["Salary"]>=70000 and row["Experience"]>=7:
        return "Top"
    elif row["Salary"]>=50000 and row["Experience"]>=3:
        return "Mid"
    else:
        return "Entry"

df["Employee_Segment"]=df.apply(emp_seg,axis=1)
df
```
