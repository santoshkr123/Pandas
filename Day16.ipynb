```python
import pandas as pd

data = {
    "Month": ["Jan", "Jan", "Jan", "Feb", "Feb", "Feb",
              "Mar", "Mar", "Mar", "Apr", "Apr", "Apr"],

    "Department": ["IT", "HR", "Finance",
                   "IT", "HR", "Finance",
                   "IT", "HR", "Finance",
                   "IT", "HR", "Finance"],

    "Sales": [50000, 30000, 40000,
              60000, 35000, 45000,
              55000, 32000, 50000,
              70000, 40000, 55000],

    "Profit": [10000, 6000, 8000,
               12000, 7000, 9000,
               11000, 6500, 10000,
               15000, 8000, 12000],

    "Employees": [10, 8, 6,
                  11, 8, 7,
                  12, 9, 7,
                  13, 9, 8]
}

df = pd.DataFrame(data)

df
```

```python
df.pivot(
    index="Month",
    columns="Department",
    values="Sales"
)
```

```python
df.pivot_table(
    index="Month",
    columns="Department",
    values="Sales",
    aggfunc="sum"
)
```

```python
df.pivot_table(
    index="Month",
    columns="Department",
    values="Sales",
    aggfunc="mean"
)
```

```python
df.pivot_table(
    index="Month",
    columns="Department",
    values="Sales",
    aggfunc="max"
)
```

```python
df.pivot_table(
    index="Month",
    columns="Department",
    values="Sales",
    aggfunc="count"
)
```

```python
df.pivot_table(
    index="Month",
    columns="Department",
    values="Sales",
    aggfunc=["sum","mean","max","count"]
)
```

```python
df.pivot_table(
    index="Month",
    columns="Department",
    values=["Sales","Profit"],
    aggfunc="sum"
)
```

```python
df.pivot_table(
    index="Month",
    columns="Department",
    values="Sales",
    aggfunc="sum",
    margins=True
)
```

```python
df.pivot_table(
    index="Month",
    columns="Department",
    values="Sales",
    aggfunc="sum",
    fill_value=0
)
```
