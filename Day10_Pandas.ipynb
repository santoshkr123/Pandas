```python
import pandas as pd

data = {
    "Order_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Month": [
        "Jan", "Jan", "Feb", "Feb", "Mar",
        "Mar", "Apr", "Apr", "May", "May"
    ],
    "Department": [
        "IT", "HR", "IT", "Finance", "HR",
        "IT", "Finance", "HR", "IT", "Finance"
    ],
    "Employee": [
        "Rahul", "Priya", "Amit", "Neha", "Rohit",
        "Sneha", "Vikas", "Pooja", "Karan", "Anjali"
    ],
    "Sales": [
        5000, 7000, 4500, 9000, 6000,
        8000, 5500, 7500, 6500, 8500
    ],
    "Profit": [
        1000, 1500, 800, 2000, 1200,
        1800, 1100, 1600, 1300, 1900
    ]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
df.pivot_table(
    values="Sales",
    index="Department",
    aggfunc="sum"
)
```

```python
#pivot by month
df.pivot_table(
    values="Sales",
    index="Month",
    aggfunc="sum"
)
```

```python
#Multiple Aggregations
df.pivot_table(
    values="Sales",
    index="Department",
    aggfunc=["sum","mean","max"]
)
```

```python
#multiple values
df.pivot_table(
    values=["Sales","Profit"],
    index="Department",
    aggfunc="sum"
)
```

```python
#columns-powerful part -Sales by Department and Month
df.pivot_table(
    values="Sales",
    index="Department",
    columns="Month",
    aggfunc="sum"
)
```

```python
#fill_value-Some Department × Month combinations don't exist.They may appear as:NaN
df.pivot_table(
    values="Sales",
    index="Department",
    columns="Month",
    aggfunc="sum",
    fill_value=0
)

```

```python
#margins=True -Want totals?-margin=grand totals#
df.pivot_table(
    values="Sales",
    index="Department",
    columns="Month",
    aggfunc="sum",
    fill_value=0,
    margins=True
)
```

```python
#Multiple Index Columns
df.pivot_table(
    values="Sales",
    index=["Department","Employee"],
    aggfunc="sum"
)
```

```python
#Multiple Dimensions
df.pivot_table(
    values="Sales",
    index="Department",
    columns="Month",
    aggfunc="sum",
    fill_value=0,
    margins=True
)
```

```python
#crosstab()-crosstab() is particularly useful for frequency/count tables.
df = pd.DataFrame({
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female"]
})
df
```

```python
#Number of employees by Department and Gender.
pd.crosstab(
    df["Department"],
    df["Gender"]
)
```

```python
#crosstab() with percentages
#Within each department, what percentage belongs to each gender?
pd.crosstab(
    df["Department"],
    df["Gender"],
    normalize="index"
)
```

```python
summary=df.groupby("Department").agg(
    **{
        "Total_Sales":("Sales","sum"),
        "Total_Profit":("Profit","sum"),
        "Average_Sales":("Sales","mean"),
        "Maximum_Sales":("Sales","max")
    }
)
summary
```
