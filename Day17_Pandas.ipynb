```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Amit", "Neha", "Rohit"],
    "Salary": [30000, 40000, 50000, 60000, 70000]
})
df
```

```python
Maan lo company salary ke basis par employees ko increment dena chahti hai:
- Salary ₹50,000 se kam hai → 10% increment.
- Salary ₹50,000 ya usse zyada hai → 5% increment.
```

```python
df["New_Salary"]=np.where(
    df["Salary"]<50000,
    df["Salary"]*1.10,
    df["Salary"]*1.05
)
df
```

```python
Q6: Salary ₹40,000 se kam ho toh "Junior", warna "Senior" assign karo. Column: Level.



Q7: Salary ₹50,000 se zyada ho toh "Eligible", warna "Not Eligible" assign karo. Column: Bonus_Status.



Q8: Salary ₹40,000 se kam ho toh 15% increment, warna 7% increment calculate karo. Column: Revised_Salary.



Q9: Salary ₹30,000 se zyada aur ₹60,000 se kam ho toh "Medium", otherwise "Other" assign karo.



Q10: Salary ₹50,000 ya usse zyada ho toh ₹5,000 bonus, warna ₹2,000 bonus do. Column: Bonus.
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Amit", "Neha", "Rohit"],
    "Salary": [30000, 40000, 50000, 60000, 70000]
})
df
```

```python
Q6: Salary ₹40,000 se kam ho toh "Junior", warna "Senior" assign karo. Column: Level.
```

```python
df["Level"]=np.where(
    df["Salary"]<40000,"Junior","Senior"
)
df
```

```python
Q7: Salary ₹50,000 se zyada ho toh "Eligible", warna "Not Eligible" assign karo. Column: Bonus_Status.
```

```python
df["Bonus_Status"]=np.where(
    df["Salary"]>50000,"Eligible","Not Eligible"
)
df
```

```python
Q8: Salary ₹40,000 se kam ho toh 15% increment, warna 7% increment calculate karo. Column: Revised_Salary.
```

```python
df["Revised_Salary"]=np.where(
    df["Salary"]<40000,
    df["Salary"]*1.15,
    df["Salary"]*1.07
)
df
```

```python
Q9: Salary ₹30,000 se zyada aur ₹60,000 se kam ho toh "Medium", otherwise "Other" assign karo.
```

```python
np.where(
    (df["Salary"]>30000) & (df["Salary"]<60000) ,"Medium","Other"
)
```

```python
Q10: Salary ₹50,000 ya usse zyada ho toh ₹5,000 bonus, warna ₹2,000 bonus do. Column: Bonus.
```

```python
df["Bonus"]=np.where(
    df["Salary"]>50000,
    df["Salary"]+5000,
    df["Salary"]+2000
)
df
```

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Amit", "Neha", "Rohit"],
    "Age": [22, 25, 28, 32, 35],
    "Salary": [30000, 40000, 50000, 60000, 70000],
    "Department": ["IT", "HR", "IT", "Finance", "HR"]
})

df.memory_usage(deep=True)
```

```python
df.memory_usage(deep=True).sum()
```

```python
megabytes me convert krne ke lye
```

```python
df.memory_usage(deep=True).sum()/(1024**2)
```

```python
df["Age"].astype('int32')
```

```python
df["Salary"].astype('int32')
```

```python
df["Department"].astype('category')
```

```python
df["Department"] = df["Department"].astype("object")
before = df["Department"].memory_usage(deep=True)

df["Department"] = df["Department"].astype("category")
after = df["Department"].memory_usage(deep=True)

print("Before:", before)
print("After:", after)
```
