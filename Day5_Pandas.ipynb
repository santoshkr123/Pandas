```python
import pandas as pd

data = {
    "Name": ["Rahul", "Priya", "Amit", "Neha", "Rohit", "Sneha"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Salary": [50000, 45000, 65000, 55000, 40000, 70000],
    "Experience": [2, 5, 7, 4, 1, 6]
}

df = pd.DataFrame(data)
df
```

```python
df.groupby("Department")["Salary"].mean()
```

```python
#Question:What is the total salary paid by each department?
df.groupby("Department")["Salary"].sum()
```

```python
#max() min()  Highest salary in each department
df.groupby("Department")["Salary"].max()
```

```python
df.groupby("Department")["Salary"].min()
```

```python
#Highest experience in each department
df.groupby("Department")["Experience"].max()
```

```python
#count()How many employees are there in each department?
df.groupby("Department")["Name"].count()
```

```python
#nunique() How many unique employees are there in each department?
df.groupby("Department")["Name"].nunique()
```

```python
df["Department"].nunique()
```

```python
#Multiple Aggregations with agg() -> For each department, find minimum salary, maximum salary, and average salary.
df.groupby("Department")["Salary"].agg(
    ["min","max","mean"]
)
```

```python
#Different Aggregations for Different Columns
'''Question:

For each department, find:

average salary
maximum salary
average experience
number of employees'''


df.groupby("Department").agg({
    "Salary" :"mean",
    "Salary" : "max",
    "Experience" :"mean",
    "Name":"count"
})
```

```python
df.groupby("Department").agg(
    Average_Salary=("Salary","mean"),
    Maximum_Salary=("Salary","max"),
    Average_Experience=("Experience","mean"),
    Number_Of_Employees=("Name","count")
    
    
    
)
```

```python
import pandas as pd

data = {
    "Name": ["Rahul", "Priya", "Amit", "Neha", "Rohit", "Sneha", "Vikas", "Pooja"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT", "Finance", "HR"],
    "Salary": [50000, 45000, 65000, 55000, 40000, 70000, 60000, 50000],
    "Experience": [2, 5, 7, 4, 1, 6, 5, 3]
}

df = pd.DataFrame(data)
df
```

```python
#Q1. Find the total salary for each department.
df.groupby("Department")["Salary"].sum()
```

```python
#Q2. Find the average salary for each department.
df.groupby("Department")["Salary"].mean()
```

```python
#Q3. Find the maximum salary for each department.
df.groupby("Department")["Salary"].max()
```

```python
#Q4. Find the minimum salary for each department.
df.groupby("Department")["Salary"].min()
```

```python
#Q5. Find the average experience for each department.
df.groupby("Department")["Experience"].mean()
```

```python
#Q6. Count the number of employees in each department.
df.groupby("Department")["Name"].count()
```

```python
#Q7. Find the number of unique employees in each department.
df.groupby("Department")["Name"].nunique()
```

```python
Q8. Find the total salary and average salary for each department.

Q9. Find minimum, maximum and average salary for each department.

Q10. Find average salary and average experience for each department.

Q11. Find employee count and total salary for each department.

Q12. Find all of these for each department:

Employee count
Total salary
Average salary
Maximum salary
Average experience

Use one groupby().agg().
```

```python
#Q8. Find the total salary and average salary for each department.
df.groupby("Department")["Salary"].agg(
    ["sum","mean"]
)
```

```python
#Q9. Find minimum, maximum and average salary for each department.
df.groupby("Department")["Salary"].agg(
    ["min","max","mean"]
)
```

```python
#Q10. Find average salary and average experience for each department.
df.groupby("Department").agg(
    Average_Salary=("Salary","mean"),
    Average_Experience=("Experience","mean")
    
    
)
```

```python
#Q11. Find employee count and total salary for each department.
df.groupby("Department").agg(
    Employee_Count=("Name","count"),
    Total_Salary=("Salary","sum") 
    
    
)
```

```python

'''Q12. Find all of these for each department:

Employee count
Total salary
Average salary
Maximum salary
Average experience

Use one groupby().agg().'''

df.groupby("Department").agg(
    Employee_Count=("Name","count"),
    Total_Salary=("Salary","sum"),
    Average_Salary=("Salary","mean"),
    Maximum_Salary=("Salary","max"),
    Average_Experience=("Experience","mean")
    
    
    
    
    
)
```

```python
#13. Which department has the highest total salary?
df.groupby("Department")["Salary"].agg(
    ["max","sum"]
)
```

```python
#Q14. Which department has the highest average salary?
df.groupby("Department")["Salary"].agg(
    ["max","mean"]
)
```

```python
#Q15. Which department has the lowest average salary?
df.groupby("Department")["Salary"].agg(
    ["min","mean"]
)
```

```python
#Q16. Find the department where employees have the highest average experience.
df.groupby("Department")["Experience"].agg(
    ["max","mean"]
)
```

```python
#Q17. Find departments having more than 2 employees.
df["Department"].value_counts() > 2
```

```python
#Q19.Find the department whose total salary expense is highest.
df.groupby("Department")["Salary"].sum().idxmax()
```

```python
#Q20.Find the department whose average salary is greater than 55,000.
df.groupby("Department")["Salary"].mean()>55000
```
