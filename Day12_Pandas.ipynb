```python
import pandas as pd

data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    
    "Name": [
        "Rahul Sharma", "Priya Singh", "Amit Kumar", "Neha Verma",
        "Rohit Singh", "Sneha Gupta", "Vikas Kumar", "Pooja Sharma",
        "Karan Singh", "Anjali Verma"
    ],
    
    "Age": [25, 32, 28, 41, 24, 35, 29, 45, 31, 27],
    
    "Department": [
        "IT", "HR", "IT", "Finance", "HR",
        "IT", "Finance", "HR", "IT", "Finance"
    ],
    
    "Salary": [
        50000, 65000, 45000, 80000, 40000,
        70000, 60000, 55000, 75000, 48000
    ],
    
    "Experience": [
        2, 6, 3, 10, 1,
        7, 5, 8, 6, 2
    ],
    
    "City": [
        "Delhi", "Mumbai", "Delhi", "Pune", "Mumbai",
        "Delhi", "Pune", "Mumbai", "Delhi", "Pune"
    ]
}

df = pd.DataFrame(data)
df
```
